Zad 1
Uzywając standardowego slownika z języka Python napisac funkcję [xi, ni]=freq(x, prob=True),
która dla zadanej kolumny danych x dyskretnych zwróci: unikalne wartości xi, ich estymowane prawdopodobienstwa pi lub częstości ni.
Zad 2
Napisac funkcj¦ [xi, yi, ni] = freq2(x,y, prob=True), która dla zadanych kolumn danych x i y zwróci:
unikalne wartości atrybutów xi, yi oraz lączny rozkalad częstości lub liczności ni (w zalezności od prarametru prob).

```python
def freq(x, y = None, prob=True):
    if y is not None:
        xy = []
        for i in range(len(x)):
            xy.append((x[i],y[i]))
        x = xy

    unique, counts = np.unique(x, axis=0, return_counts=True)

    if prob:
        probability = counts / len(x)
        return (unique, probability)

    return (unique, counts)
```

```python
# Przykladowe dane
x = np.array([1, 1, 1, 1, 2, 2, 3, 3])
y = np.array(['a', 'b', 'b', 'a', 'b', 'b', 'a', 'a'])
```

```python
print(freq(x, prob=True))
print(freq(x, prob=False))

print(freq(x, y, prob=True))
print(freq(x, y, prob=False))
```

Zad 3
Wykorzystując powyzsze funkcje, napisac funkcje, które wyliczą:
entropię h=entropy(x) oraz przyrost informacji i=infogain(x,y):

```python
def entropy(x, y=None):
    _, probs = freq(x, y)

    sum = 0
    for prob in probs:
        sum += -prob * np.log2(prob)

    return sum
```

```python
print(entropy(x))
print(entropy(y))
print(entropy(x, y))
```

```python
def infogain(x, y):
    return entropy(x) + entropy(y) - entropy(x, y)
```

```python
infogain(x, y)
```

Zad 4
Wczytac dane testowe zoo.csv oraz dokonac selekcji/stopniowania atrybutów z wykorzystaniem
kryterium przyrostu informacji.

```python
dane = pd.read_csv("zoo.csv")

typeCol = dane["type"]

res = []
for name, col in dane.items():
    if name != "animal" and name != "type":
        res.append((name, entropy(col), infogain(col, typeCol)))

resPd = pd.DataFrame(res, columns=["cecha", "entropy", "infogain"])
resPd = resPd.sort_values(by="infogain", ascending=False)
print(resPd)
```
